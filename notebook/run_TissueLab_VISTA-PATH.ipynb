# Applying Active Learning Segmentation with VISTA-PATH in TissueLab
  
**Project:** VISTA-PATH: An interactive foundation model for pathology image segmentation and quantitative analysis in computational pathology  

## Overview

This tutorial demonstrates how to apply VISTA-PATH within the TissueLab platform to perform interactive whole-slide image (WSI) segmentation using an active learning workflow.

By the end of this tutorial, you will be able to:

1. Load whole-slide images (WSIs)

2. Configure and run workflows using the MUSK model and VISTA-PATH

3. Perform interactive segmentation with human-in-the-loop refinement



## Step 1: Log In to TissueLab

Begin by logging in to the **TissueLab** online platform:

## Step 2: Load Your Whole-Slide Images (WSIs)

From the **Dashboard**, upload your own whole-slide images (WSIs) to begin the analysis.

1. Click **Upload Files**.
2. Select the desired slide file from your local machine.
3. Click **Image Viewer** to open the slide. Confirm that it loads correctly and that zooming and panning function smoothly.

  <img src="../figures/upload_file.png" width="700" alt="Uploading a whole-slide image in TissueLab">

> **Note:** Supported file formats may include `.svs`, `.ndpi`, and `.tiff`. Processing time may vary depending on slide size and network speed.

## Step 3: Get Patch Embedding

In this step, we generate patch embeddings using the MUSK model to prepare for subsequent patch-level annotation.

1. Click **Workflow** in the right panel. 
2. Under **Tissue Classification**, select **MUSK Classification**.
3. Once the workflow panel appears, click **Run Workflow**. You may adjust the patch size and WSI level to control how patches are extracted. The default setting uses a patch size of 224 at level 0 (highest resolution).
4. Wait for the workflow to finish. Once completed, press the **X** shortcut key to display the extracted patches.

## Step 4: Perform Patch-Level Annotations

Refine the patch-level classification results by annotating representative patch samples. In this example, we demonstrate the workflow using the **tumor** class; the same procedure can be applied to other tissue classes.

1. Select the **Polygon** or **Rectangle** annotation tool from the top panel.  
2. Create **negative annotations** (e.g., *Negative Control*) on patches that do not contain tumor tissue.  
3. Create **positive annotations** (e.g., *Tumor*) on patches that contain tumor tissue.  
4. The classifier will automatically update its predictions based on your annotations. You can repeat this process iteratively until the results are satisfactory.
